```python
counts = df["diseases"].value_counts()

print("Diseases with 1-9 samples:", ((counts >= 1) & (counts <= 9)).sum())
print("Diseases with 10-29 samples:", ((counts >= 10) & (counts <= 29)).sum())
print("Diseases with 30-49 samples:", ((counts >= 30) & (counts <= 49)).sum())
print("Diseases with 50-99 samples:", ((counts >= 50) & (counts <= 99)).sum())
print("Diseases with 100-199 samples:", ((counts >= 100) & (counts <= 199)).sum())
print("Diseases with 200-499 samples:", ((counts >= 200) & (counts <= 499)).sum())
print("Diseases with 500+ samples:", (counts >= 500).sum())
```

```python
min_samples=50
disease_count=df["diseases"].value_counts()
valid_diseases=disease_count[disease_count>=min_samples].index
df_filtered = df[df["diseases"].isin(valid_diseases)]
print("Original rows:", len(df))
print("Filtered rows:", len(df_filtered))
print("Original diseases:", df["diseases"].nunique())
print("Filtered diseases:", df_filtered["diseases"].nunique())
```

```python
df_filtered.iloc[:, 1:].stack().value_counts()
```

```python
df_filtered.isnull().sum().sum()
```

```python
df_filtered.duplicated().sum()
```

```python
symptom_columns = df_filtered.columns[1:]

duplicate_groups = (
    df_filtered
    .groupby(list(symptom_columns))["diseases"]
    .nunique()
)

conflicting_groups = duplicate_groups[duplicate_groups > 1]

print("Duplicate symptom patterns:", len(duplicate_groups))
print("Conflicting symptom patterns:", len(conflicting_groups))
```

```python

conflict_rows = (
    df_filtered
    .merge(
        conflicting_groups.rename("disease_count"),
        left_on=list(symptom_columns),
        right_index=True
    )
)

print("Rows belonging to conflicting patterns:", len(conflict_rows))
print("Percentage of dataset:", round(len(conflict_rows) / len(df_filtered) * 100, 2), "%")
```

```python
conflicting_groups.value_counts().sort_index()
```

```python
df_clean = df_filtered.drop_duplicates()

print("Rows before removing duplicates:", len(df_filtered))
print("Rows after removing duplicates:", len(df_clean))
print("Rows removed:", len(df_filtered) - len(df_clean))
print("Diseases remaining:", df_clean["diseases"].nunique())
```
